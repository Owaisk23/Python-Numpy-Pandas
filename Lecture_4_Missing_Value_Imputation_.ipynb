***Missing data refers to the absence of values for certain observations and is an unavoidable problem in most data sources. ***

Scikit-learn does not support missing values as input, so we need to remove observations with missing data or transform them into permitted values.

***The act of replacing missing data with statistical estimates of missing values is called imputation. ***

The goal of any imputation technique is to produce a complete dataset that can be used to train machine learning models. There are multiple imputation
techniques we can apply to our data. The choice of imputation technique we use will depend on
1. whether the data is missing at random,
2. the number of missing values, and
3. the machine learning model we intend to use.

Technique # 1 :**Removing observations with missing data**

Complete Case Analysis (CCA), also called list-wise deletion of cases, consists
of discarding those observations where the values in any of the variables are missing.

1.   CCA can be applied to categorical and numerical variables.
2.   CCA is quick and easy to implement
3.   If data is missing across many variables, CCA may lead to the removal of a big portion of the dataset

Technique # 2: **Mean or Median Imputation**

Mean or median imputation consists of **replacing missing values with the variable mean or median**. This ***can only be performed in numerical variables***.

The mean or the median is *calculated using a train set*, and these values are used to impute missing data in train and test sets, as well as in future data we intend to score with the machine learning model. Therefore, we need to *store these mean and median values*.

Use scikit-learn and Feature-engine libraries and pandas for value imputation.

**Use mean if variables are normally distributed and median
imputation otherwise. Mean and median imputation may distort the
distribution of the original variables if there is a high percentage of
missing data**

Technique # 3: **Replacing Missing Value with an arbitrary number**
Arbitrary number imputation consists of replacing missing values with an arbitrary value.Some commonly used values include 999, 9999, or -1 for positive distributions.

1.  This method is suitable for numerical variables.
2.  When replacing missing values with an arbitrary number, we need to be careful not to select a value close to the mean or the median, or any other common value of the distribution.

Important:
Arbitrary number imputation can be used when data is   ** not missing at random**, when we are **building non-linear models**, and when the **percentage of missing data is high**.

**This imputation technique distorts the original variable distribution.**

Technique # 4: **Replace missing value with a value at the end of the distribution**

Replacing missing values with a value at the end of the variable distribution is equivalent to replacing them with an arbitrary value, but instead of identifying the arbitrary values manually, these values are automatically selected as those at the very end of the variable distribution.

The values that are used to replace missing information are *estimated using
the mean plus or minus three times the standard deviation if the variable is normally distributed, or the inter-quartile range (IQR) proximity rule otherwise.*

According to the IQR proximity rule, missing values will be replaced with the 75th quantile + (IQR * 1.5) at the right tail or by the 25th quantile - (IQR * 1.5) at the left tail. The IQR is given by the 75th quantile - the 25th quantile.

**End-of-tail imputation may distort the distribution of the original
variables, so it may not be suitable for linear models.**

image.png

```
from feature_engine.imputation import EndTailImputer
gaussimputer = EndTailImputer(tail='right', variables=['LotFrontage'])
IQRimputer = EndTailImputer(imputation_method= 'iqr', tail='right', fold= 1.5, variables=['LotFrontage'])
maximputer = EndTailImputer(imputation_method= 'max', variables=['LotFrontage'])
```

```
gaussimputer.fit(X_train)
gaussimputer.imputer_dict_
```

```
IQRimputer.fit(X_train)
IQRimputer.imputer_dict_
```

```
maximputer.fit(X_train)
maximputer.imputer_dict_
```

```
X_train_gauss = gaussimputer.transform(X_train)
X_test_gauss = gaussimputer.transform(X_test)
```

```
X_train_iqr = IQRimputer.transform(X_train)
X_test_iqr = IQRimputer.transform(X_test)
```

```
X_train_max = maximputer.transform(X_train)
X_test_max = maximputer.transform(X_test)
```

```
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20,7))
X_train['LotFrontage'].plot(kind='kde', ax= ax1)
X_train_gauss['LotFrontage'].plot(kind='kde', ax=ax1, color='red')
X_train['LotFrontage'].plot(kind='kde', ax=ax2, color='blue')
X_train_iqr['LotFrontage'].plot(kind='kde', ax=ax2, color='red')
X_train['LotFrontage'].plot(kind='kde', ax=ax3, color='blue')
X_train_max['LotFrontage'].plot(kind='kde', ax=ax3, color='red')
```

Technique # 6: **Radom Sample Imputation**

Random sampling imputation consists of extracting random observations from the pool of available values in the variable.

**Random sampling imputation preserves the original distribution, and is suitable for numerical and categorical variables alike**.

**If the variables indicated in the random_state list are not numerical the imputer will return an error. The variables indicated as seed should not contain missing values.**

```
from feature_engine.imputation import RandomSampleImputer
imputerGeneral = RandomSampleImputer()
```

```
imputerGeneral.fit(X_train)
X_train_general = imputerGeneral.transform(X_train)
X_test_general = imputerGeneral.transform(X_test)
```

```
#fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20,7))
X_train['LotFrontage'].plot(kind='kde')
X_train_general['LotFrontage'].plot(kind='kde', color='red')
```
