***ADD a categorical third variable***

A common modification of the basic scatter plot is the addition of a third variable. Values of the third variable can be encoded by modifying how the points are plotted. For a third variable that indicates categorical values (like geographical region or gender), the most common encoding is through point color. Giving each point a distinct hue makes it easy to show membership of each point to a respective group.



```python
df['Species'].unique()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# df is your DataFrame containing the Iris dataset
x = df['PetalLengthCm']
y = df['SepalLengthCm']
category = df['Species']  # 'Species' is the categorical variable

# Create a color map for different categories
color_map = {'Iris-setosa': 'red', 'Iris-versicolor': 'green', 'Iris-virginica': 'blue'}

#This line maps each category in the 'Species' column (stored in the variable category) to its corresponding color defined
#in the color_map dictionary.
#The map() function is used to apply the mapping defined in the color_map dictionary to each category in the 'Species' column.
colors = category.map(color_map)

# Plotting the scatter plot with color-coded categories
plt.scatter(x, y, c=colors, label='Data')

# Adding labels and title
plt.xlabel('Petal Length (cm)')
plt.ylabel('Sepal Length (cm)')
plt.title('IRIS DATASET Scatter Plot with Trend Line')

# Displaying the legend
plt.legend()

# Showing the plot
plt.show()
```
